## 16. Load Benchmark Summary Table

```python
import pandas as pd
from pathlib import Path

rows = []

for runtime_file in Path("outputs/phase4").glob("*/runtime.json"):
    run_dir = runtime_file.parent

    predictions = pd.read_csv(run_dir / "predictions.csv")

    rows.append({
        "run": run_dir.name,
        "n_samples": len(predictions),
        "accuracy": predictions["correct"].mean(),
        "invalid_rate": predictions["predicted_author"].isna().mean(),
        "unique_predictions": predictions["predicted_author"].nunique(dropna=True),
    })

summary = pd.DataFrame(rows)
summary = summary.sort_values("accuracy", ascending=False)

summary
```

## 17. Inspect Parsed Status Distribution

```python
import pandas as pd

runs = [
    "mistral_zero_shot_seed42",
    "mistral_few_shot_seed42",
    "mistral_cot_zero_shot_seed42",
    "llama_zero_shot_seed42",
    "llama_few_shot_seed42",
    "llama_cot_zero_shot_seed42",
]

for run in runs:
    path = f"outputs/phase4/{run}/predictions.csv"

    try:
        df = pd.read_csv(path)

        print("=" * 80)
        print(run)
        print("=" * 80)

        print(df["parsed_status"].value_counts(dropna=False))
        print()

    except FileNotFoundError:
        print(f"Missing run: {run}")
```

## 18. Classification Report Example

```python
from sklearn.metrics import classification_report
import pandas as pd

run_name = "mistral_few_shot_seed42"

path = f"outputs/phase4/{run_name}/predictions.csv"

df = pd.read_csv(path)

clean_df = df.dropna(subset=["predicted_author"])

print(
    classification_report(
        clean_df["true_author"],
        clean_df["predicted_author"],
        zero_division=0,
    )
)
```
